```python
df["is_fraud"]=df["is_fraud"].replace(1,"True")
df["is_fraud"]=df["is_fraud"].replace(0,"False")
df
```

```python
from sklearn.preprocessing import LabelEncoder
coder=LabelEncoder()
```

```python
df["is_fraud"]=df["is_fraud"].astype(str)
df["is_fraud"]=coder.fit_transform(df["is_fraud"])
df
```

```python
df["amount"]=df["amount"].astype(str)
df["amount"]=coder.fit_transform(df["amount"])
df
```

```python
df["merchant_category"]=df["merchant_category"].astype(str)
df["merchant_category"]=coder.fit_transform(df["merchant_category"])
df
```
